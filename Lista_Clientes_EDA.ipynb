```python
df['gender'].value_counts().sort_values(ascending=True).plot(
    kind="bar",
    title="Distribución por género de los Clientes",
    figsize=(10, 6),
    color='mediumturquoise' 
);
plt.xlabel("Género del Cliente")
plt.ylabel("Número de Clientes")
plt.show()
```

```python
sns.kdeplot(df['income'], fill=True, color="lightsalmon")
plt.title("Distribución del ingreso de los Clientes")
plt.xlabel("Frecuencia")
```

El gráfico de densidad (Kernel Density Estimate) presenta la distribución del ingreso de los clientes, donde el eje horizontal muestra la frecuencia o el ingreso, y el eje vertical indica la densidad de probabilidad. La característica más notable de esta distribución es su naturaleza bimodal, lo que se evidencia en los dos picos prominentes de la curva: uno situado aproximadamente alrededor de $5 \times 10^6$ (5 millones) y el segundo, más alto, cerca de $10 \times 10^6$ (10 millones). Esta forma sugiere fuertemente que la población de clientes se compone de dos grupos o segmentos distintos, cada uno con un nivel de ingreso predominante diferente. La mayor concentración de clientes se encuentra en el rango de ingresos entre el primer y el segundo pico, con la distribución extendiéndose hasta aproximadamente $20 \times 10^6$ (20 millones) antes de que la densidad caiga a cero.

```python
sns.kdeplot(df['age'], fill=True, color="palevioletred")
plt.title("Distribución de la Edad de los Clientes")
plt.xlabel("Frecuencia")
```

El gráfico de densidad que ilustra la distribución de la edad de los clientes presenta una curva notablemente ancha y relativamente estable en el eje vertical, lo que indica que la clientela está distribuida de manera sólida a lo largo de un rango de edad extenso, principalmente entre los 20 y 65 años. Esta forma sugiere que la empresa ha logrado captar y retener la atención de diversas generaciones o cohortes de edad, manteniendo una base de clientes robusta tanto en el segmento de adultos jóvenes como en el de la mediana edad.

```python
sns.kdeplot(df['educ'], fill=True, color="lightcoral")
plt.title("Distribución de los Años de Educación de los Clientes")
plt.xlabel("Frecuencia")
```

El gráfico de densidad que representa la distribución de los años de educación de los clientes también exhibe una forma multimodal y una amplia dispersión, reflejando una diversidad en el nivel educativo de la base de clientes. En el eje horizontal, que probablemente representa el número de años de educación formal, la curva se concentra en un rango que va aproximadamente desde los 2 hasta los 16 años. Esta estructura con múltiples modas sugiere que existen varios grupos educativos importantes en la clientela, como podría ser un grupo con educación primaria/secundaria parcial y otro grupo sustancial con educación secundaria completa o incluso con inicios de educación superior, indicando que la base de clientes no se limita a un único nivel de formación.

```python
# Gráfico de torta de años de educación por género
for gen in df['gender'].unique():
    education_counts = df[df['gender'] == gen]['educ'].value_counts().sort_index()
    
    plt.figure(figsize=(8, 8))
    plt.pie(education_counts,
            labels=education_counts.index,  # muestra los años exactos
            autopct='%1.1f%%',
            startangle=90,
            colors=plt.cm.tab20.colors)  # colores automáticos para muchas categorías
    plt.title(f'Distribución de años de educación - {gen}')
    plt.axis('equal')
    plt.show()
```

Los gráficos circulares ilustran la distribución de los años de educación (donde cada número representa un año de estudio) segmentada por género, mostrando que, aunque la clientela tiene una diversidad educativa general, existen diferencias significativas en la concentración de niveles. En la población Femenina, los años de educación con mayor proporción son el 1 (8.4%) y el 7 (7.7%), seguido de cerca por el año 2 (7.7%), lo que sugiere un número considerable de mujeres con un nivel educativo bajo o primario. Por otro lado, la población Masculina muestra una distribución más fuerte en los niveles intermedios, donde el nivel más alto corresponde a los 11 años (10.2%), seguido por los 2 años (8.6%) y el 10 (7.5%), lo cual indica que los clientes masculinos están sobrerrepresentados en el nivel que podría corresponder a la finalización de la educación secundaria. En general, los hombres muestran una mayor concentración en el nivel secundario completo (11 años), mientras que las mujeres se concentran ligeramente más en niveles educativos inferiores.

```python
# Histograma de años de educación por género
for gen in df['gender'].unique():
    data = df[df['gender'] == gen]['educ']
    
    plt.figure(figsize=(8, 6))
    plt.hist(data, bins=range(df['educ'].min(), df['educ'].max() + 2), edgecolor='black', color='lightskyblue')
    plt.title(f'Distribución de años de educación - {gen}')
    plt.xlabel('Años de educación')
    plt.ylabel('Cantidad de personas')
    plt.grid(axis='y', alpha=0.75)
    plt.show()
```

Los histogramas de la distribución de los años de educación confirman y detallan los patrones observados previamente al comparar la cantidad absoluta de clientes en cada nivel. En la población Masculina, la distribución es más extrema y puntiaguda, destacando fuertemente el pico de 12 años de educación, con una cantidad de personas cercana a 45, siendo este el nivel educativo más común. Otros picos notables en hombres se encuentran alrededor de los 3, 6, 9 y 14 años, sugiriendo múltiples niveles educativos de finalización importantes. En contraste, la población Femenina presenta una distribución más dispersa y uniforme, aunque el nivel más alto de concentración de clientes se da en los 1, 8 y 9 años de educación, con la cantidad de personas bordeando las 35. Las mujeres tienen una mayor cantidad de clientes en el nivel más bajo (1 año) que los hombres, pero los hombres demuestran una mayor frecuencia en el nivel de finalización de la escuela secundaria (12 años), lo que resulta en una base de clientes masculina con una mayor concentración en el nivel educativo de secundaria completa.

```python
variables_numericas = df[['age', 'educ', 'income']]
rosa_paleta = ["#FFEDF0", "#FFC0CB", "#FFB6C1"]
correlacion = variables_numericas.corr()
plt.figure(figsize=(8, 7))  
sns.heatmap(correlacion, annot=True, cmap=rosa_paleta, vmin=-1, vmax=1, square=True, fmt=".2f")
plt.title('Mapa de Calor de Correlación entre Variables Numéricas de Clientes', fontsize=12)
plt.show()
```

El mapa de calor de correlación muestra la fuerza y dirección de la relación lineal entre cada par de variables, utilizando el Coeficiente de Correlación de Pearson, donde 1.00 significa correlación perfecta positiva, -1.00 significa correlación perfecta negativa, y 0.00 significa ausencia de correlación lineal. Como era de esperarse, la correlación de una variable consigo misma (la diagonal) es 1.00. Sin embargo, las correlaciones cruzadas son consistentemente extremadamente cercanas a cero: la correlación entre edad e ingreso es -0.04, la correlación entre educación e ingreso es -0.01, y la correlación entre edad y educación es 0.01. Estos valores confirman de manera concluyente que no existe una relación lineal significativa entre la edad, los años de educación y el ingreso de los clientes. Los tres factores son, por lo tanto, variables independientes al considerar su relación lineal dentro de esta base de datos.

Conclusiones Clave del Análisis de la Clientela

Las variables de ingreso y edad, aunque no están correlacionadas linealmente entre sí, presentan distribuciones multimodales que sugieren la existencia de segmentos internos:

Ingreso: La distribución del ingreso es bimodal (dos picos principales alrededor de 5 y 10 millones). Esto indica que la base de clientes está compuesta por dos grupos económicos distintos que deberían ser objetivo de estrategias de marketing o productos diferenciados.

Edad: La distribución de la edad es multimodal y ancha (picos alrededor de 30 y 45-50 años). Esto demuestra que la clientela es diversa y no se enfoca en una sola generación, sino que cubre sólidamente los segmentos de adultos jóvenes y de mediana edad.

El análisis de educación desagregado por género reveló diferencias clave en la composición educativa de la clientela:

Mujeres: Tienen una concentración de clientes en los niveles educativos bajos (1, 2 y 7 años), sugiriendo una mayor presencia en el segmento de educación primaria o secundaria parcial.

Hombres: Muestran la mayor concentración en el nivel de 12 años de educación, que a menudo corresponde a la finalización de la escuela secundaria. Esto podría indicar que los productos o servicios de la compañía atraen a hombres en la etapa de transición post-secundaria.

Los diagramas de dispersión y el mapa de calor de correlación (con valores cercanos a cero) confirmaron el hallazgo más importante para el modelado predictivo:

Independencia Total: El Ingreso es completamente independiente de la Edad y de los Años de Educación. En esta base de clientes, ser joven o mayor, o tener mucha o poca educación, no predice si un cliente tendrá un ingreso alto o bajo.

Dado que las variables demográficas clave (Edad, Educación) no tienen una relación lineal con la variable económica principal (Ingreso), las conclusiones son:

Segmentación por Distribución, No por Correlación: Las estrategias de negocio deben aprovechar las distribuciones multimodales (los picos) para definir segmentos, en lugar de buscar relaciones predictivas simples. Por ejemplo, se deben crear dos campañas dirigidas a los dos grupos de ingresos distintos.

Marketing Específico por Género/Educación: La mayor disparidad se encuentra en los niveles educativos por género. Es esencial diseñar productos o comunicaciones diferentes para el segmento de hombres con 12 años de educación frente al segmento de mujeres con menos de 8 años de educación.

Búsqueda de Otras Variables: Si se quiere predecir el Ingreso o el Comportamiento del Cliente, la compañía debe buscar otros factores, ya que las variables demográficas básicas son ineficaces. Se deben analizar variables de comportamiento (transacciones, uso de productos, antigüedad, ubicación, etc.) para encontrar una correlación que permita la predicción.
